```python
from fredapi import Fred
import pandas as pd
import matplotlib.pyplot as plt

# Fetch economic indicator data from FRED
def fetch_economic_indicator(api_key, series_id, start_date):
    fred = Fred(api_key=api_key)
    data = fred.get_series(series_id, start=start_date)
    return data

# Visualize economic indicators
def plot_economic_indicator(data, series_id):
    data.plot(figsize=(10, 6), color='green')
    plt.title(f"{series_id} Economic Indicator")
    plt.xlabel("Date")
    plt.ylabel("Value")
    plt.grid()
    plt.show()

# Example usage
api_key = 'YOUR_FRED_API_KEY'  # Replace with your FRED API key
series_id = 'GDP'  # Gross Domestic Product
start_date = '2015-01-01'
economic_data = fetch_economic_indicator(api_key, series_id, start_date)
plot_economic_indicator(economic_data, series_id)
```
