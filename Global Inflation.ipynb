```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# Fetch global inflation data
def fetch_inflation_data(country_code):
    url = f"http://api.worldbank.org/v2/country/{country_code}/indicator/FP.CPI.TOTL?format=json"
    response = requests.get(url)
    data = response.json()[1]
    df = pd.DataFrame(data)
    df = df[['date', 'value']].dropna()
    df['date'] = pd.to_datetime(df['date'])
    return df

# Visualize inflation trends
def plot_inflation_trends(data, country_code):
    plt.figure(figsize=(10, 6))
    plt.plot(data['date'], data['value'], label=f"{country_code} Inflation Rate", color='red')
    plt.title(f"Inflation Rate Trends for {country_code}")
    plt.xlabel("Year")
    plt.ylabel("Inflation Rate (%)")
    plt.legend()
    plt.grid()
    plt.show()

# Example usage
country_code = 'US'  # United States
inflation_data = fetch_inflation_data(country_code)
plot_inflation_trends(inflation_data, country_code)
```
