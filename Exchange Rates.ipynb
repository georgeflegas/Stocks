```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# Fetch historical exchange rates
def fetch_exchange_rates(app_id, base_currency, start_date, end_date):
    url = f"https://openexchangerates.org/api/historical.json"
    params = {
        'app_id': app_id,
        'start': start_date,
        'end': end_date,
        'base': base_currency
    }
    response = requests.get(url, params=params)
    data = response.json()
    df = pd.DataFrame(data)
    return df

# Visualize exchange rates
def plot_exchange_rates(data, currency):
    plt.figure(figsize=(10, 6))
    plt.plot(data.index, data[currency], label=f"Exchange Rate: {currency}/{data.base_currency}", color='orange')
    plt.title(f"Exchange Rate Over Time")
    plt.xlabel("Date")
    plt.ylabel("Rate")
    plt.legend()
    plt.grid()
    plt.show()

# Example usage
app_id = 'YOUR_APP_ID'  # Replace with your Open Exchange Rates App ID
base_currency = 'USD'
currency = 'EUR'
start_date = '2023-01-01'
end_date = '2023-10-01'
exchange_data = fetch_exchange_rates(app_id, base_currency, start_date, end_date)
plot_exchange_rates(exchange_data, currency)
```
