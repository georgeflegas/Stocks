```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# Fetch cryptocurrency data
def fetch_crypto_data(symbol, currency, limit):
    url = f"https://min-api.cryptocompare.com/data/v2/histoday"
    params = {
        'fsym': symbol,
        'tsym': currency,
        'limit': limit,
        'api_key': 'YOUR_API_KEY'  # Replace with your CryptoCompare API key
    }
    response = requests.get(url, params=params)
    data = response.json()['Data']['Data']
    df = pd.DataFrame(data)
    df['time'] = pd.to_datetime(df['time'], unit='s')
    return df

# Visualize cryptocurrency prices
def plot_crypto_prices(data, symbol):
    plt.figure(figsize=(10, 6))
    plt.plot(data['time'], data['close'], label=f"{symbol} Closing Price", color='purple')
    plt.title(f"{symbol} Cryptocurrency Price Over Time")
    plt.xlabel("Date")
    plt.ylabel("Price")
    plt.legend()
    plt.grid()
    plt.show()

# Example usage
symbol = 'BTC'  # Bitcoin
currency = 'USD'
limit = 365  # Last 365 days
crypto_data = fetch_crypto_data(symbol, currency, limit)
plot_crypto_prices(crypto_data, symbol)
```
